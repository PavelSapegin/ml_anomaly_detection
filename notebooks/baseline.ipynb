```python
import numpy as np
import pandas as pd
from sklearn.metrics import average_precision_score
```

```python
X_val = pd.read_csv("../data/processed/X_val.csv")
y_val = pd.read_csv("../data/processed/y_val.csv")
```

```python
cols_with_high_corr = ["V17", "V14", "V12", "V10"]

anomaly_score = X_val[cols_with_high_corr].abs().sum(axis=1)
```

```python
y_val_series = y_val.squeeze()

anomaly_score_class_0 = anomaly_score[y_val_series == 0]
anomaly_score_class_1 = anomaly_score[y_val_series == 1]
```

```python
print(anomaly_score_class_0.describe())
print(anomaly_score_class_1.describe())
```

```python
pr_auc = average_precision_score(y_val, anomaly_score)
pr_auc
```
