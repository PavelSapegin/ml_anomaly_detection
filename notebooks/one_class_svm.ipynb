```python
import sys
from pathlib import Path

import joblib
import pandas as pd
from sklearn.svm import OneClassSVM

sys.path.append(str(Path.cwd().parent))

from src.evaluation import evaluate
```

```python
X_train, X_val = [pd.read_csv(f'../data/processed/X_{split}.csv') for split in
                          ['train', 'val']]
y_train, y_val = [pd.read_csv(
    f'../data/processed/y_{split}.csv').squeeze("columns")
                          for split in ['train', 'val']]
```

```python
model = OneClassSVM(kernel="rbf", gamma="auto", nu=0.05)


X_train_subset = X_train.sample(n=45_000, random_state=42)
model.fit(X_train_subset)
evaluate(model, X_val, y_val)
```

```python
output_dir = Path("../models")
output_dir.mkdir(parents=True, exist_ok=True)
joblib.dump(model, "../models/one_class_svm.joblib")
```

```python
model = OneClassSVM(kernel="rbf", gamma="auto")

cols_with_high_corr = ["V17", "V14", "V12", "V10"]
X_train_subset = X_train.sample(n=45_000, random_state=42)
model.fit(X_train_subset[cols_with_high_corr])
evaluate(model, X_val[cols_with_high_corr], y_val)
```
