```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/creditcard.csv")
print(df.shape)
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
print(f"Duplicates: {df.duplicated().sum()}")
print(f"Misses: {df.isna().sum()}")
```

```python
extra_copies = df[df.duplicated()]["Class"].value_counts()
extra_copies
```

```python
cols = df.columns.to_list()
cols.remove("Class")

conflicting_groups = df.groupby(cols)["Class"].nunique()
conflicting_groups = conflicting_groups[conflicting_groups > 1]

print(f"Количество строк/групп с расхождениями: {len(conflicting_groups)}")
```

```python
df = df.drop_duplicates()
```

```python
sns.countplot(data=df,x="Class")
plt.yscale("log")
plt.show()
counts = df["Class"].value_counts()
percentages = df["Class"].value_counts(normalize=True).mul(100).round(2)
summary= pd.concat([counts, percentages], axis = 1, keys=["Count", "Percent"])

print(summary)
```

```python
print(df.groupby("Class")["Amount"].describe())
```

```python
sns.boxenplot(data=df, x="Class",y="Amount")
plt.yscale("log")
plt.show()
```

```python
df["Hours"] = (df["Time"] % 86_400) / 3600

hours_class_0 = df[df["Class"] == 0]["Hours"]
hours_class_1 = df[df["Class"] == 1]["Hours"]

fraud_rate = df.groupby(df["Hours"].astype(int))["Class"].mean().reset_index()

fig, axes = plt.subplots(1,3, figsize=(12,6))

sns.histplot(hours_class_0, ax=axes[0], kde=True, stat="density", bins=24)
axes[0].set_title("Distribution Hours by Class 0")

sns.histplot(hours_class_1, ax=axes[1], kde=True, stat="density",bins=24)
axes[1].set_title("Distribution Hours by Class 1")

sns.lineplot(data=fraud_rate, x="Hours", y="Class")
axes[2].set_title("Fraud rate by hours")

plt.tight_layout()
plt.show()
```

```python
corr_matrix = df.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=False, fmt=".2f", cmap="coolwarm")

plt.title("Matrix of features correlation")
plt.show()
```

```python
correlations = df.corr()["Class"].sort_values(ascending=False)

print(correlations.drop("Class"))
```

```python
fig, axes = plt.subplots(1,2,figsize=(8,6))

sns.boxenplot(data=df, x="Class", y="V17", ax=axes[0])
axes[0].set_title("Boxplot of V17 by Class")

sns.boxenplot(data=df, x="Class", y="V14", ax=axes[1])
axes[1].set_title("Boxplot of V14 by Class")

plt.show()
```

```python
df = df.drop(columns=["Hours"])
df.to_csv("../data/processed/creditcard_clean.csv",index=False)
```
